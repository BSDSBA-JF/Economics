```python
import os
import shutil
import pandas as pd
import numpy as np
import pyarrow as pa
import pyarrow.parquet as pq
```

```python
source_folder = '../lfs'
parquet_folder = os.path.join('data', 'parquet')
excel_folder = os.path.join('data', 'excel')

# Make sure subfolders exist
os.makedirs(parquet_folder, exist_ok=True)
os.makedirs(excel_folder, exist_ok=True)

# List all LFS folders/files containing 'PHL'
lfs_files = [x for x in os.listdir(source_folder) if 'PHL' in x]

for lfs_file in lfs_files:
    print("Folder:", lfs_file)
    lfs_file_path = os.path.join(source_folder, lfs_file)
    
    elements_lfs_file = os.listdir(lfs_file_path)

    for element in elements_lfs_file:
        source_path = os.path.join(lfs_file_path, element)

        if element.lower().endswith('.csv'):
            # Read CSV survey data
            df = pd.read_csv(source_path, low_memory=False)
            df = df.replace('  ', np.nan).copy()
            
            # Convert to Parquet
            table = pa.Table.from_pandas(df)
            parquet_filename = os.path.splitext(element)[0] + '.parquet'
            parquet_path = os.path.join(parquet_folder, parquet_filename)
            pq.write_table(table, parquet_path)

        elif element.lower().endswith(('.xls', '.xlsx')):
            # Copy Excel data dictionary into excel subfolder
            dest_path = os.path.join(excel_folder, element)
            shutil.copy2(source_path, dest_path)
```
